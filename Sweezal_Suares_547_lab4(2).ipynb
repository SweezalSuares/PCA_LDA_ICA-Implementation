```python
# Define target variables explicitly
y = cancer.target
target_names = cancer.target_names
df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_cancer['target'] = y
df_cancer['Diagnosis'] = [target_names[i] for i in y]
```

```python
#distribution count
df_cancer.hist(figsize=(20,20))
plt.show()
```

```python
plt.figure(figsize=(7, 5))
ax = sns.countplot(data=df_cancer, x='Diagnosis', hue='Diagnosis', palette='Set2', legend=False)
plt.title('Class Distribution: Breast Cancer Wisconsin Dataset')
plt.xlabel('Diagnosis')
plt.ylabel('Count')
plt.show()
```

```python
# Perform data preprocessing using only the 30 numeric feature columns
from sklearn.preprocessing import StandardScaler

features_to_scale = df_cancer.drop(columns=['target', 'Diagnosis'], errors='ignore')
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features_to_scale)
```

```python
#Apply PCA to reduce the dimensionality of the dataset
from sklearn.decomposition import PCA
pca = PCA(n_components=4)
X_pca = pca.fit_transform(X_scaled)
display(X_pca)
```

```python
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
print(cumulative_variance)
```

```python
n90=np.argmax(cumulative_variance>=0.90)+1
n95=np.argmax(cumulative_variance>=0.95)+1
print(f"Number of components needed for 90% variance: {n90}")
print(f"Number of components needed for 95% variance: {n95}")
```

```python
# Re-create PCA DataFrame
component_names = ['PC 1', 'PC 2', 'PC 3', 'PC 4']
pca_df = pd.DataFrame(data=X_pca, columns=component_names)
pca_df['Target'] = y
pca_df['Diagnosis'] = [target_names[i] for i in y]
print(pca_df.head().to_string(index=False))
```

```python
exp_var_ratio = pca.explained_variance_ratio_ * 100
cum_var_ratio = np.cumsum(exp_var_ratio)
plt.figure(figsize=(10, 6))
bars = plt.bar(component_names, exp_var_ratio, color='#377eb8', alpha=0.7, label='Individual Variance')
plt.plot(component_names, cum_var_ratio, color='#ff7f00', linestyle='--', marker='o', label='Cumulative Variance')
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 1, f'{yval:.1f}%', ha='center', va='bottom', fontsize=10)
plt.title('Variance Explained by 4 Principal Components', fontsize=14, pad=15, fontweight='bold')
plt.xlabel('Principal Components', fontsize=12)
plt.ylabel('Percentage of Variance (%)', fontsize=12)
plt.ylim(0, 110)
plt.legend(loc='center right')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Apply LDA and project data onto 1 component
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X_scaled, y)
lda_df = pd.DataFrame(data=X_lda, columns=['LDA Component 1'])
lda_df['Target'] = y
lda_df['Diagnosis'] = [target_names[i] for i in y]
print(lda_df.head().to_string(index=False))
```

```python
sns.histplot(data=lda_df[lda_df['Target'] == 0], x='LDA Component 1',
             color='#e41a1c', label='Malignant', kde=True, alpha=0.6, bins=20)
sns.histplot(data=lda_df[lda_df['Target'] == 1], x='LDA Component 1',
             color='#377eb8', label='Benign', kde=True, alpha=0.6, bins=20)
plt.title('LDA Projection: Optimal 1D Class Separation', fontsize=14, pad=15, fontweight='bold')
plt.xlabel('Linear Discriminant 1', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(title="Diagnosis")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python

# Apply ICA and construct DataFrame
from sklearn.decomposition import FastICA

ica = FastICA(n_components=4, random_state=42, max_iter=1000)
X_ica = ica.fit_transform(X_scaled)

ic_names = ['IC 1', 'IC 2', 'IC 3', 'IC 4']
ica_df = pd.DataFrame(data=X_ica, columns=ic_names)
ica_df['Target'] = y
ica_df['Diagnosis'] = [target_names[i] for i in y]
print(ica_df.head().to_string(index=False))
```

```python
sns.set_theme(style="ticks")
g = sns.pairplot(ica_df, vars=ic_names, hue="Diagnosis",
                 palette={'malignant': '#e41a1c', 'benign': '#377eb8'},
                 plot_kws={'alpha': 0.6, 's': 30, 'edgecolor': 'none'},
                 diag_kind='kde',
                 diag_kws={'alpha': 0.5})

g.fig.suptitle('ICA Projections: Pairwise Relationships of 4 Independent Components',
               y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=pca_df, x='PC 1', y='PC 2', hue='Diagnosis', palette={'malignant': '#e41a1c', 'benign': '#377eb8'}, alpha=0.7)
plt.title('PCA: PC 1 vs PC 2', fontsize=12, fontweight='bold')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
np.random.seed(42)
jitter = np.random.normal(0, 0.1, size=len(lda_df))
sns.scatterplot(x=lda_df['LDA Component 1'], y=jitter, hue=lda_df['Diagnosis'], palette={'malignant': '#e41a1c', 'benign': '#377eb8'}, alpha=0.7)
plt.title('LDA: Component 1 with Jitter', fontsize=12, fontweight='bold')
plt.xlabel('Linear Discriminant 1')
plt.ylabel('Jitter')
plt.ylim(-0.5, 0.5)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=ica_df, x='IC 1', y='IC 2', hue='Diagnosis', palette={'malignant': '#e41a1c', 'benign': '#377eb8'}, alpha=0.7)
plt.title('ICA: IC 1 vs IC 2', fontsize=12, fontweight='bold')
plt.xlabel('Independent Component 1')
plt.ylabel('Independent Component 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Apply t-SNE to the Dataset

```python
# Apply t-SNE
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000)
X_tsne = tsne.fit_transform(X_scaled)

tsne_df = pd.DataFrame(data=X_tsne, columns=['t-SNE Component 1', 't-SNE Component 2'])
tsne_df['Target'] = y
tsne_df['Diagnosis'] = [target_names[i] for i in y]
display(tsne_df.head())
```

### Comprehensive Comparison: PCA vs LDA vs ICA vs t-SNE

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Initialize a 2x2 grid of subplots for all four methods
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()
pal = {'malignant': '#e41a1c', 'benign': '#377eb8'}

# 1. PCA Visualization
sns.scatterplot(ax=axes[0], data=pca_df, x='PC 1', y='PC 2', hue='Diagnosis', palette=pal, alpha=0.7)
axes[0].set_title('PCA: PC 1 vs PC 2', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. LDA Visualization (1D with Jitter)
np.random.seed(42)
jitter = np.random.normal(0, 0.1, size=len(lda_df))
sns.scatterplot(ax=axes[1], x=lda_df['LDA Component 1'], y=jitter, hue=lda_df['Diagnosis'], palette=pal, alpha=0.7)
axes[1].set_title('LDA: Component 1 with Jitter', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Linear Discriminant 1')
axes[1].set_ylabel('Jitter')
axes[1].set_ylim(-0.5, 0.5)
axes[1].grid(True, linestyle='--', alpha=0.5)

# 3. ICA Visualization
sns.scatterplot(ax=axes[2], data=ica_df, x='IC 1', y='IC 2', hue='Diagnosis', palette=pal, alpha=0.7)
axes[2].set_title('ICA: IC 1 vs IC 2', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Independent Component 1')
axes[2].set_ylabel('Independent Component 2')
axes[2].grid(True, linestyle='--', alpha=0.5)

# 4. t-SNE Visualization
sns.scatterplot(ax=axes[3], data=tsne_df, x='t-SNE Component 1', y='t-SNE Component 2', hue='Diagnosis', palette=pal, alpha=0.7)
axes[3].set_title('t-SNE: Dimension 1 vs Dimension 2', fontsize=12, fontweight='bold')
axes[3].set_xlabel('t-SNE Component 1')
axes[3].set_ylabel('t-SNE Component 2')
axes[3].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('Comparison of Dimensionality Reduction Transformations', fontsize=16, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```
